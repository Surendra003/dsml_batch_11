Project Title:

Student Management System Using OOP Concepts

Problem Statement:

Develop a Student Management System using Python that demonstrates
the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction.

The system should allow users to store student information, calculate
grades, and display student details based on different student categories.

Requirements

Create an Abstract Class Student

The abstract class should contain:

Student ID

Student Name

A private attribute
__
marks

A method to set marks

A method to get marks

An abstract method calculate
_grade()

A method display_
details()

```python
from abc import ABC, abstractmethod
# Abstract Class
class Student(ABC):
    def __init__(self, student_id, student_name):
        self.student_id = student_id
        self.student_name = student_name
        self.__marks = 0
    def set_marks(self, marks):
        self.__marks = marks
    def get_marks(self):
        return self.__marks
    @abstractmethod
    def calculate_grade(self):
        pass
    def display_details(self):
        print("\n----- Student Details -----")
        print("ID:", self.student_id)
        print("Name:", self.student_name)
        print("Marks:", self.get_marks())
class ScienceStudent(Student):
    def calculate_grade(self):
        marks = self.get_marks()
        if marks >= 80:
            return "A"
        elif marks >= 60:
            return "B"
        elif marks >= 40:
            return "C"
        else:
            return "F"
class ManagementStudent(Student):
    def calculate_grade(self):
        marks = self.get_marks()
        if marks >= 75:
            return "A"
        elif marks >= 55:
            return "B"
        elif marks >= 35:
            return "C"
        else:
            return "F"
category = input("Enter Category (Science/Management): ")
student_id = int(input("Enter Student ID: "))
student_name = input("Enter Student Name: ")
marks = float(input("Enter Marks: "))
if category.lower() == "science":
    student = ScienceStudent(student_id, student_name)
elif category.lower() == "management":
    student = ManagementStudent(student_id, student_name)
else:
    print("Invalid Category")
student.set_marks(marks)
student.display_details()
print("Category:", category)
print("Grade:", student.calculate_grade())
```
