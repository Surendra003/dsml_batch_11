Generate grades for each student (like in SEE)

```python
Students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
Marks = [45, 67, 98, 87, 76, 88]

def student_result(students, marks):

    result = {}

    for student, mark in zip(students, marks):

        if 100 >= mark >= 90:
            grade = 'A+'
        elif 90 > mark >= 80:
            grade = 'A'
        elif 80 > mark >= 70:
            grade = 'B+'
        elif 70 > mark >= 60:
            grade = 'B'
        else:
            grade = 'Non Grade'

        result[student] = {
            "Marks": mark,
            "Grade": grade
        }

    return result

print(student_result(Students, Marks))

```

Create a list containing only the students who have passed the
examination. (marks above 40 is pass)

```python
passed = [(student, mark) for student, mark in zip(Students, Marks) if mark > 40]

print(passed)
```

Calculate the total marks obtained by all students combined.

```python
result = {}
for student, mark in zip(Students, Marks):
    result[student] = mark
total = sum(result.values())

print("Total Marks:", total)
```
