```python
print("Menu")
print("Type 1 for Balance enquiry")
print("Type 2 for Deposite")
print("Type 3 for exit")
balance = 23000
num=int(input("Enter a number:"))
if num==1:
  print(f"Your balance is: {balance}")
  exit = int(input("Press 3 to exit "))
  while exit==3:
    break
if num==2:
  deposite = int(input("Enter a amount you want to deposite: "))
  new_balance = balance+deposite
  print(f"You have successively deposited Nrs.{deposite} and your current balance is Nrs.{new_balance}")
  exit = int(input("Press 3 to exit "))
  while exit==3:
    break
while num==3:
  break

```
