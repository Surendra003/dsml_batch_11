```python
#class task
class BankAccount:
  def __init__(self,name,initial_balance = 0):
    self.name=name
    self.initial_balance=initial_balance
  def user_info(self):
    print(f"Your name is {self.name}")
  def deposite(self,amount):
    if amount>0:
      self.add=int(input(f"{self.name} how much balance you want to deposite: "))
      print(f"You have successfully deposited Rs.{self.add}")
      print(f"Your new amount is {amount+self.add}")
  def withdraw(self,amount):
    self.withd = int(input(f"{self.name} how much balance you want to withdraw: "))
    if amount>0:
      print(f"You have successfully withdraw Rs.{self.withd}")
      print(f"Your new amount is {amount-self.withd}")
    else:
      print("You have Insufficient balance!!")
  def check(self):
    print(f"{self.name} has {self.initial_balance}")
```

```python
class SavingAccount(BankAccount):
  def add_interest(self, rate=5):
    initial_balance = self.check()
    interest = (initial_balance*rate)/100
    print(f"Interest {interest}")
```
