```python
#Inheritance
#Single Inheritance
class Animal:
  def sound(self):
    print("animal make sound")

class Dog(Animal):
  def bark(self):
    print("Dog bark")
obj3 = Dog()
obj3.sound()
obj3.bark()
```

```python
#Multiple Inheritance
class Father:
  def blood(self):
    print("O+")
class Mother:
  def laugh(self):
    print("better smile")
class child(Father,Mother):
  def play(self):
    print("Play football")
obj4 = child()
obj4.blood()
```

```python
class Father:
  def blood(self):
    print("O+")
class Mother:
  def laugh(self):
    print("better smile")
class child(Father,Mother):
  pass
obj4 = child()
obj4.laugh()
```

```python
#Multilevel Inheritance
class GrandFather:
  def blood(self):
    print("O+")
class Father(GrandFather):
  def skill(self):
    print("Have general skill")
class child(Father):
  def play(self):
    print("Liked Play football")
obj4 = child()
obj4.skill()
obj4.blood()
```

```python
#Hierarchical Inheritance
class Parent:
  def blood(self):
    print("B+")
class boy(Parent):
  def skills(self):
    print("Futsal")
class girl(Parent):
  def skill(self):
    print("Like dancing")
obj5 = girl()
obj5.skill()
obj5.blood()
obj6 = Parent()
obj6.blood()
```

```python
#Hybrid Inheritance
class Person:
  def name(self):
    print("Surendra")
  def age(self):
    return 33
class Teacher(Person):
  def subject(self):
    print("Statistics")
class Student(Person):
  def student_id(self):
    print("101")
class TeachingAssistant(Student,Teacher):
  def department(self):
    print("Science Dept")
  def salary(self):
    return '40000'
obj7 = TeachingAssistant()
print(f"Age is : {obj7.age()}")
```

```python
class OTTPsubscription:
  def __init__(self, subscription_id, plan, total_payment):
    self.id = subscription_id
    self.plan = plan
    self.total_payment = total_payment

  def subscription(self):
    print(f"Subscriber with {self.id} id subscribed to the {self.plan}")
  def unsubscribe(self):
    print(f"Subscriber with {self.id} unsubscribed to the {self.plan}")
```

```python
class PremiumSubscription(OTTPsubscription):
  def __init__(self, subscription_id, plan, total_payment, screens):
    super().__init__(subscription_id, plan, total_payment)
    self.max_screens = screens
  def set_max_screen(self,screens):
    self.max_screens = screens
    print(f"maximum screen set to {self.max_screens} in the premium plan")
```

```python
obj = PremiumSubscription(122,"pro",12000,7)
obj.set_max_screen(6)
```

```python
#Class Task
class Person:
  def __init__(self, name, age, **kwargs):
    self.name = name
    self.age = age
    super().__init__(**kwargs)
  def person_info(self):
    print(f"Name is {self.name} and his/her age is {self.age}")
class Teacher(Person):
  def __init__(self,name,age,subject,**kwargs):
    super().__init__(name, age,**kwargs)
    self.subject=subject
    print(f"Teacher named {self.name} teaches {self.subject}")
obj= Teacher("hari",23,"Science")
```

```python
class Student(Person):
  def __init__(self,name, age, id,**kwargs):
    super().__init__(name,age,**kwargs)
    self.id=id
    print(f"{self.name} id is {self.id}")
```

```python
class TeachingAssistant(Student,Teacher):
  def __init__(self,name,age,subject,department,salary):
    super().__init__(name,age,subject)
    self.department=department
    self.salary=salary
    print(f"Teaching Assistant named {self.name}, age:{self.age} is in {self.department} and his salary is {self.salary}")
```
