**`Students grade management system`**

Create a Python class named Student.


Requirements:


1.Create a class attribute named school.


2.Use the __init__() method to take:


Student name


Student marks


3.Create an instance method named check_grade() that displays:


A Grade: Marks ≥ 80


B Grade: Marks ≥ 60 and < 80


C Grade: Marks ≥ 40 and < 60


Fail: Marks < 40


4.Create a class method to display the school name.


5.Create a static method to display a welcome message.


6.Create an object and display the student's grade.

```python
class Student:
  school_name = "Gyan Niketan"
  def __init__(self):
    self.sname = input("Enter Student name: ")
    self.smarks = float(input(f"Enter {self.sname} marks: "))
  def check_grade(self):
    if 100>=self.smarks>=80:
      print("A")
    elif 80>=self.smarks>=60:
      print("B")
    elif 60>=self.smarks>=40:
      print("C")
    elif self.smarks<40:
      print("Fail")
    else:
      print("Invalid")
  def school(cls):
    print(f"School Name: {cls.school_name}")
  @staticmethod
  def welcome_message():
    print("Welcome to the school.")
student_data = Student()
student_data.welcome_message()
student_data.check_grade()
student_data.school()
```
