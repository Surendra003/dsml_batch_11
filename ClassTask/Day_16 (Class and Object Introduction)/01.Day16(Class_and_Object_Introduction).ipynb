```python
class Student:
  '''Student Information'''
  School = "Skill Shikshya" #Class Attribute
  def __init__(self, name, age):
    self.name = name  #Instant Attribute
    self.age = age
  @classmethod
  def get_school(cls):  #class method
    print(cls.School)
s1 = Student("Surendra",22)
s1.name
s1.age
Student.get_school()
```

```python
class Calculator:
  @staticmethod
  def add(a,b):
    return a+b
print(Calculator.add(3,4))
```

```python
from os import name
class Dog:
  count = 0 #class attribute

  def __init__(self, name, age=3):
    self.name = name
    self.age = age
    Dog.count+=1
  def bark(self):
    print(f"{self.name} is barking")
  @classmethod
  def get_count(cls):
    return cls.count

  @staticmethod
  def get_vet_ingo():
    return f"the doctor of {name} is Dr. Ram K.C"
```

```python
a = Dog("sunu")
print(a.age)
print(a.count)
print(a.bark())
print(a.get_count())
print(a.get_vet_ingo())
```

```python
class Calculators:
  name = "Skill Shikshya"
  def __init__(self):
    self.numbers1=int(input("Enter a first numbers:"))
    self.numbers2=int(input("Enter a second numbers:"))
    self.operator= input("Enter a operator:")
  def calculate(self):
    if self.operator=="+":
      print(self.numbers1+self.numbers2)
    elif self.operator=="-":
      print(self.numbers1-self.numbers2)
    elif self.operator=="*":
      print(self.numbers1*self.numbers2)
    elif self.operator=="/":
      print(self.numbers1/self.numbers2)
    else:
      print("Invalid!!")
  @classmethod
  def show_company(cls):
    print("Company",cls.name)
  @staticmethod
  def welcome():
    print("Welcome to Calculator.")
Calculators.welcome()
a = Calculators()
a.calculate()
a.welcome()
```
