```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
factorial(5)
```

```python
def recur_fib(n):
  if n<=1:
    return n
  else:
    return (recur_fib(n-1)+recur_fib(n-2))
nterms = 10
for i in range(nterms):
  print(recur_fib(i))
```

```python
def outer(x):
  def inner(y):
    return x+y
  return inner
```

```python
#function as an argument
def add(x,y):
  return x +y
def calculate(func,x,y):
  return func(x,y)
result= calculate(add,4,6)
result
```

```python
#square example
def square(x):
  return x*x
def calculate(func,x):
  return func(x)
result = calculate(square,6)
result
```

```python
#function as a value
def greeting(name):
  def hello():
    return "hello, "+name+"!"
  return hello
greet = greeting("Dhiraj")
print(greet())
```

```python
#Decorator
def myWrapper(func):
  def myFunc():
    print("Inside Wrapper!")
    func()
  return myFunc
```

```python
@myWrapper
def myFunc():
  print("Hello, world")
myFunc()
```

```python
def logger(func):
  def wrapper(*args,**kwargs):
    print(f"Calling {func.__name__}")
    return func(*args,**kwargs)
  return  wrapper
```

```python
@logger
def greet(name,age):
  print(f"Hello, {name}, Age:{age}")
greet("Surendra",33)
```

```python
def logger1(func):
  def wrapper(*args):
    return func(*args)
  return wrapper
```

```python
@logger1
def hello(name):
  print(f"Hello, {name}")
hello('Dhiraj')
```
