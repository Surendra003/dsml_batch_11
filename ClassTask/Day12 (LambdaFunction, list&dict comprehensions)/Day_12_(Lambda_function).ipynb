```python
num = [1,2,3,4,5]
list(map(lambda num: num**2,num))
```

```python
from functools import reduce
num = [1,2,3,4,5]
list(filter(lambda num: num%2==0,num))
res = reduce(lambda x,y: x*y,num)
print(res)
```

```python
number = float(input("Enter a number:"))
result = lambda number: "even" if number%2==0 else "odd"
result(number)
```

```python
#list.sort(reserve = True|False, key = lambda x:x[1])
students = [
    ("Ram",80),
    ("Sita",70)
]
students.sort(key = lambda x:x[1])
print(students)
```

```python
students = [
    ("Ram",80),
    ("Sita",70)
]
students.sort(reverse = True, key = lambda x:x[1])
print(students)
```

```python
employee = [
    {"name":"Ram","Salary":5000},
    {"name":"Hari","Salary":40000},
    {"name":"Sitas","Salary":45000}

]
employee.sort(reverse = True, key=lambda emp:emp["Salary"])
print(employee)
```

```python
list1 = [1,2]
list2 = ["A","B"]
pairs = [(x,y) for x in list1 for y in list2]
print(pairs)
```
