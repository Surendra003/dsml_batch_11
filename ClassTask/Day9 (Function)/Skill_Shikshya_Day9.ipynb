```python
def calculator(number1,number2,sign):
  if sign=="+":
    return number1+number2
  elif sign=="-":
    return number1-number2
  elif sign=="*":
    return number1*number2
  elif sign=="/":
    return number1/number2
number1 = int(input("Enter a First Number: "))
number2 = int(input("Enter a Second Number: "))
sign = input("Enter a operation")
calculator(number1,number2,sign)
```

```python
#make a empty list, integer append, if float then pop
lst=[]
for number in range(5):
  num=float(input("Enter a five number:"))
  if num==int(num):
    lst.append(int(num))
    print(lst)
  elif num==float(num):
    lst.pop()
    print(lst)
  else:
    print("!!!Invalid number")
```
