Write a function that takes three numbers as arguments and returns the largest number

```python
def largest_number(a,b,c):
  return max(a,b,c)
largest_number(10,20,30)
```

```python
def vowel(text):
  vowels = 'aeiouAEIOU'
  count=0

  if text in vowels:
    count+=1
  return count
vowel("Hi guys")
```

```python
def count_vowel(text):
  vowels="aeiouAEIOU"
  count=0
  for char in text:
    if char in vowels:
      count+=1
  return count

count_vowel("Hello Nepal")
```
