```python
def multiply(*args):
  result = 1
  for num in args:
    result*=num
  return result
print(multiply(2,3,4,5))
```

```python
def fun(**kwargs):
  for k, val in kwargs.items():
    print(k, '=',val)
fun(s='Python',s1='is',s2='Awesome')
```

```python
def introduce(**kwargs):
  details = []
  for k, v in kwargs.items():
    details.append(k+":"+str(v))
  return ','.join(details)
print(introduce(Name="Surendra",Age="24"))
```

purpose of the filter() function is to filter out the required output

```python
def even(x):
  if x%2==0:
    return x
list3 = [1,2,3,4,5,6,7]
even_number=filter(even,list3)
print(list(even_number))
```

```python
def odd(x):
  if x%2!=0:
    return x
list4 = [1,2,3,4,5,6,34,55]
odd_number = filter(odd,list4)
print(list(odd_number))
```

```python
from functools import reduce
def lesser(x,y):
  return x*y
list6 = []
number_of_product = int(input("Enter a product number your want to add: "))
for i in range(number_of_product):
  products = int(input(f"Enter {i+1} Products: "))
  list6.append(products)
print(list6)
res=reduce(lesser,list6)
print(res)
```
